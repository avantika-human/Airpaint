testing and running mediapipe


#importing libraries


```python
import cv2
import mediapipe as mp
```

```python
handsm = mp.solutions.hands
drawing = mp.solutions.drawing_utils
```

```python
cap = cv2.VideoCapture(0)
```

```python
with handsm.Hands(
    max_num_hands = 1,
    min_detection_confidence = 0.7, #detection is true only if the model is 70% confident
    min_tracking_confidence = 0.7 #detection is true only if the model is 70% confident
) as hands:
    while(True):
        ret,frame = cap.read() #ret means it gives True or false if the frame is given a new frame
        if not ret:
            break
        frame = cv2.flip(frame, 1) #flipping the frame so that it shows to you like a mirror
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) #mediapipe expects color in RGB but cv2 takes and gives color is BGR so convert the color
        result = hands.process(rgb)
        if result.multi_hand_landmarks:
            for hand in result.multi_hand_landmarks:
                drawing.draw_landmarks(frame, hand, handsm.HAND_CONNECTIONS)

                h,w,_ = frame.shape
                fingertip = hand.landmark[8]
                x,y = int(fingertip.x *w), int(fingertip.y *h)

                cv2.circle(frame, (x,y), 12, (241,218,165), -1)
            
        cv2.imshow("Hand Tracking", frame)
        if (cv2.waitKey(1) & 0xFF == 27):
                break
cap.release()
cv2.destroyAllWindows()
```
